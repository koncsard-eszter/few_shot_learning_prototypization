```julia
# xompute euclidean distances between queries and prototypes
function euclidean_dist(queries, prototypes)
    # queries: (D, Q), prototypes: (D, N)
    # Returns matrix of size (N, Q)
    num_of_prototypes = size(prototypes, 2)
    num_of_queries = size(queries, 2)
    
    dists = zeros(Float32, n_prototypes, n_queries)
    for i in 1:num_of_prototypes
        for j in 1:num_of_queries
            dists[i, j] = sum((prototypes[:, i] .- queries[:, j]).^2)
        end
    end
    return dists
end
```
